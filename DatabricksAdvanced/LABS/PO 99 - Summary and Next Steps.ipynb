##### EXPAND FOR ADDITIONAL NOTES

<details>

#### The Complete Optimization Toolkit

This visual shows the three pillars of performance optimization you explored across all demos and labs. Each builds on the others to deliver fast, efficient workloads.

#### Data Layout

How your data is physically organized on disk has a direct impact on query speed. You learned that:

- **Over-partitioning** creates thousands of small files, leading to excessive cloud storage requests and slow queries. Letting Spark handle file layout avoids this problem entirely.
- **ZORDER** colocates related data in the same files, enabling data skipping for filtered queries — but only on the Z-ordered column.
- **Liquid Clustering** replaces both partitioning and ZORDER with a flexible, multi-column approach. It supports redefining clustering keys without rewriting data and delivers strong file pruning across multiple query patterns.

#### Query Execution

Even with good data layout, query execution strategy matters:

- **Shuffle joins** redistribute data across nodes, which is expensive for large tables. **Broadcast joins** send small tables to all nodes, avoiding the shuffle entirely.
- **Exploding joins** caused by duplicate keys can multiply row counts by 100x, causing massive spill to disk. Reordering joins (smaller first) and using `ANALYZE TABLE` to enable the cost-based optimizer can eliminate spill.
- **Python UDFs** bypass Spark's optimizer and process data serially by default. SQL UDFs are optimized by Catalyst and Photon. When Python UDFs are necessary, repartitioning enables parallel execution.

#### Diagnose & Optimize

The Spark UI is your primary diagnostic tool. Throughout this course you learned to read:

- **Cloud storage request counts** and **response sizes** to understand I/O efficiency
- **Files read** and **files pruned** to verify data skipping effectiveness
- **Shuffle read/write sizes** to identify expensive data movement
- **Disk spill metrics** to detect memory pressure
- **Query plans (DAGs)** to trace the full execution path and verify optimization effects

</details>
